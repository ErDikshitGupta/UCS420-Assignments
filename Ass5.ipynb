```python
# Q1)
import numpy as np
arr = np.array([10,20,30,40,50])
print("Original Array: " , arr)
print("Adding 2 in all the elements: ", arr+2)
print("Multiplying 3 with all the elements: ", arr*3)
print("Divide every element by 2: ", arr/2)
```

```python
# Q2) b)
# i)
x = np.array([1,2,3,4,5,1,2,1,1,1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("\nFor x:")
print("Most frequent value:", most_frequent)
print("Indices:", indices)
# ii)
y = np.array([1,1,1,2,3,4,2,4,3,3])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("\nFor x:")
print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
# Q4)
import numpy as np
Dikshit = np.linspace(10, 100, 25)
print("Array:", Dikshit)
print("Dimensions:", Dikshit.ndim)
print("Shape:", Dikshit.shape)
print("Total elements:", Dikshit.size)
print("Data type:", Dikshit.dtype)
print("Total bytes:", Dikshit.nbytes)
Dikshit_transpose = Dikshit.reshape(25, 1)
print("Transpose using reshape():")
print(Dikshit_transpose)
print("Transpose using T:")
print(Dikshit.T)
```

```python
# Q5)
ucs420_Dikshit = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35])
print("Original Array:")
print(ucs420_Dikshit)
print("Mean:", np.mean(ucs420_Dikshit))
print("Median:", np.median(ucs420_Dikshit))
print("Maximum:", np.max(ucs420_Dikshit))
print("Minimum:", np.min(ucs420_Dikshit))
print("Unique elements:", np.unique(ucs420_Dikshit))
reshaped_ucs420_Dikshit = ucs420_Dikshit.reshape(4, 3)
print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_Dikshit)
resized_ucs420_Dikshit = np.resize(ucs420_Dikshit, (2, 3))
print("\nResized Array (2 x 3):")
print(resized_ucs420_Dikshit)
```
