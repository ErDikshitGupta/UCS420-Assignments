```python
#Q1)
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

# b. Convert Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)

# c. Identify readings greater than 32°C
above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", above_32)

# d. Count readings exceeding 32°C
count = np.sum(corrected_temperature > 32)
print("Number of readings greater than 32°C:", count)
```

```python
#Q2)
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
total_steps = np.sum(steps)
print("Total steps:", total_steps)

# b. Mean number of steps
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)

# c. Maximum and minimum values
maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum steps:", maximum)
print("Minimum steps:", minimum)

# d. Total steps for each day (axis=0)
day_totals = np.sum(steps, axis=0)
print("Total steps for each day:", day_totals)

# e. Total steps for each user (axis=1)
user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)

# f. Position containing maximum number of steps
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum value:", max_position)
```

```python
#Q3)
import numpy as np

# a. Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original array:", original)


# b. Create a slice from index 1 to 4
subset = original[1:5]
print("Subset:", subset)


# c. Modify the first element of subset
subset[0] = 999

print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)


# d. Create another slice using copy()
copied = original[1:5].copy()

# Modify first element of copied array
copied[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied:", copied)


# e. Create numbers from 1 to 12 and reshape into 3x4
matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)


# f. Indexing and slicing

# First row
first_row = matrix[0]
print("\nFirst row:", first_row)

# Last row
last_row = matrix[-1]
print("Last row:", last_row)

# Second column
second_column = matrix[:, 1]
print("Second column:", second_column)

# Rows 1-2 and columns 2-3
selected_elements = matrix[1:3, 2:4]
print("Rows 1-2 and columns 2-3:")
print(selected_elements)


# g. Flatten using flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()

print("\nUsing flatten():", flattened)
print("Using ravel():", raveled)


# h. Modify ravel() array
raveled[0] = 1000

print("\nAfter modifying ravel:")
print("Ravel:", raveled)
print("Original matrix:")
print(matrix)


# i. Modify flatten() array
flattened[1] = 2000

print("\nAfter modifying flatten:")
print("Flatten:", flattened)
print("Original matrix:")
print(matrix)


# j. Display array properties
print("\nArray properties:")
print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
#Q4)
import numpy as np

# Given data
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])


# a. Display shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)


# b. Transpose of X
X_transpose = X.T

print("\nTranspose of X:")
print(X_transpose)

print("\nExplanation:")
print("The transpose converts rows into columns and columns into rows.")


# c. Calculate X.T @ X
XtX = X.T @ X

print("\nX.T @ X:")
print(XtX)


# d. Calculate matrix inverse
XtX_inverse = np.linalg.inv(XtX)

print("\nInverse of X.T @ X:")
print(XtX_inverse)


# e. Calculate OLS coefficients
beta = XtX_inverse @ X.T @ y

print("\nOLS coefficients:")
print(beta)


# f. Interpretation
print("\nCoefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])


# g. Prediction for a new user
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("\nPredicted assistance score:", predicted_score)
```
