```python
rn= "1024170274"
L=[]
for digit in rn:
    L.append(int(digit)*10)
print(L)
L.append(100)
print(L)
L.insert(2,16)
print(L)
L.remove(20)
print(L)
L.pop(10)
print(L)
L.sort()
print(L)
L.sort(reverse=True)
print(L)
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]
print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

```python
scores = tuple(L[:8])
print("Scores:", scores)
# i. Highest score and its index
highest = max(scores)
highest_index = scores.index(highest)
# Lowest score and number of times it appears
lowest = min(scores)
lowest_count = scores.count(lowest)
print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest appears:", lowest_count)

# ii. Reverse the tuple and return it as a list
reversed_scores = list(reversed(scores))
print("Reversed tuple as list:", reversed_scores)

# iii. Ask the user for a score
score = int(input("Enter a score: "))
if score in scores:
    print("Score found at index:", scores.index(score))
else:
    print("Score is not present in the tuple.")

# iv. Attempt to change an element
scores[0] = 100

# v. Unpack the tuple
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
import random

# Set random seed using your roll number
random.seed(1024170274)

# i. Generate a list of 100 random numbers between 100 and 900
L = []
for i in range(100):
    L.append(random.randint(100, 900))
print("Random numbers:", L)

# ii. Count and print all odd numbers
odd_numbers = [x for x in L if x % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Number of odd numbers:", len(odd_numbers))

# iii. Count and print all even numbers
even_numbers = [x for x in L if x % 2 == 0]
print("Even numbers:", even_numbers)
print("Number of even numbers:", len(even_numbers))

# iv. Count and print all prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_numbers = [x for x in L if is_prime(x)]
print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

# v. Find the number occurring most frequently
most_frequent = max(set(L), key=L.count)
print("Most frequent number:", most_frequent)
print("Number of times it occurs:", L.count(most_frequent))
```

```python
# First 8 digits of roll number
digits = [1, 0, 2, 4, 1, 7, 0, 2]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)

# vi. Union of A and B
print("Union:", A.union(B))

# vii. Intersection of A and B
print("Intersection:", A.intersection(B))

# viii. Difference
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference() gives values present in one set but not the other.
# Unlike symmetric_difference(), it is performed separately in each direction.

# ix. Symmetric difference
print("Symmetric difference:", A.symmetric_difference(B))

# x. Check subset and superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. Ask user for a value and discard it from A
X = int(input("Enter a value to remove from A: "))
A.discard(X)
print("Set A after discard:", A)
# discard() is safer because it does not give an error if X is not present.
```

```python
my_dict = {
    "name": "Your Name",
    "roll_no": "1024170274",
    "branch": "CSE",
    "age": 20,
    "city": "Your City"
}

print("Original dictionary:", my_dict)


# i. Rename "city" to "location"
value = my_dict.pop("city")
my_dict["location"] = value

print("After renaming city:", my_dict)


# ii. Add CGPA
my_dict["cgpa"] = 8.5

print("After adding CGPA:", my_dict)


# iii. Increase age by 1
my_dict["age"] = my_dict["age"] + 1

print("After increasing age:", my_dict)


# iv. Delete "branch" using pop() and del
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

dict_pop.pop("branch")
del dict_del["branch"]

print("Using pop():", dict_pop)
print("Using del:", dict_del)
# pop() removes the key and returns its value, while del only removes it.


# v. Iterate using items()
for key, value in my_dict.items():
    print(key, "→", value)


# vi. Check whether "email" exists
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present.")


# vii. Create second dictionary
friend_dict = {
    "name": "Friend",
    "roll_no": "1024170000",
    "branch": "ECE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:", merged_dict)
# If both dictionaries have the same key, the value from friend_dict wins.


# viii. Dictionary comprehension
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Only string values:", string_values)
```
