```python
#Q1)

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """My favorite topic is sports because it keeps me active and healthy.
I especially enjoy swimming because it improves my fitness and helps me relax.
Cricket is another sport I love because it requires teamwork, patience, and strategy.
I also enjoy football because it is exciting and full of energy.
Playing and watching these sports makes me feel happy and motivated.
I believe sports teach us discipline, confidence, and the importance of teamwork."""

# 1. Convert text to lowercase and remove punctuation using re
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("1. Cleaned Text:")
print(clean_text)

# 2. Tokenize text into words and sentences
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)
print("\n2. Word Tokens:")
print(words)
print("\nSentence Tokens:")
print(sentences)

# 3. Compare split() and word_tokenize()
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)
print("\n3. Using split():")
print(split_words)
print("\nUsing word_tokenize():")
print(nltk_words)

# 4. Remove stopwords using NLTK
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in nltk_words
    if word not in stop_words and word.isalpha()
]
print("\n4. Words after removing stopwords:")
print(filtered_words)

# 5. Display word frequency distribution excluding stopwords
frequency = FreqDist(filtered_words)
print("\n5. Word Frequency Distribution:")
for word, count in frequency.most_common():
    print(f"{word}: {count}")
```

```python
#Q2)

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """My favorite topic is sports because it keeps me active and healthy.
I especially enjoy swimming because it improves my fitness and helps me relax.
Cricket is another sport I love because it requires teamwork, patience, and strategy.
I also enjoy football because it is exciting and full of energy.
Playing and watching these sports makes me feel happy and motivated.
I believe sports teach us discipline, confidence, and the importance of teamwork."""

# 1. Extract all alphabetic words using re.findall()
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print("1. Alphabetic words:")
print(words)

# 2. Remove stopwords using NLTK's stopword list
stop_words = set(stopwords.words('english'))
filtered = [w for w in words if w not in stop_words]
print("\n2. After stopword removal:")
print(filtered)

# 3. Stemming with PorterStemmer
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered]
print("\n3. Stemmed words:")
print(stemmed)

# 4. Lemmatization with WordNetLemmatizer (default pos = noun)
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered]
print("\n4. Lemmatized words:")
print(lemmatized)

# 5. Compare stemmed vs lemmatized (only where they differ)
print("\n5. Comparison (Original -> Stemmed | Lemmatized):")
print(f"{'Original':<14}{'Stemmed':<14}{'Lemmatized':<14}")
seen = set()
for o, s, l in zip(filtered, stemmed, lemmatized):
    if o not in seen and (s != l or o != l):
        print(f"{o:<14}{s:<14}{l:<14}")
    seen.add(o)
```

```python
#Q3)

import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Three short texts of my own
texts = [
    "Central bank raises interest rates to control rising inflation across the economy",
    "Local football team wins championship after dramatic penalty shootout victory",
    "New smartphone features faster processor and improved camera for night photography"
]

# 1. Bag of Words using CountVectorizer
count_vec = CountVectorizer(stop_words='english')
bow = count_vec.fit_transform(texts)

bow_df = pd.DataFrame(bow.toarray(),
                      columns=count_vec.get_feature_names_out(),
                      index=['Text 1', 'Text 2', 'Text 3'])
print("=== Bag of Words (CountVectorizer) ===")
print(bow_df)

# 2. TF-IDF using TfidfVectorizer
tfidf_vec = TfidfVectorizer(stop_words='english')
tfidf = tfidf_vec.fit_transform(texts)

tfidf_df = pd.DataFrame(tfidf.toarray(),
                        columns=tfidf_vec.get_feature_names_out(),
                        index=['Text 1', 'Text 2', 'Text 3'])
print("\n=== TF-IDF Scores ===")
print(tfidf_df.round(3))

# 3. Top 3 keywords per text using TF-IDF
feature_names = tfidf_vec.get_feature_names_out()
print("\n=== Top 3 Keywords per Text ===")
for i, row in enumerate(tfidf.toarray()):
    top_idx = np.argsort(row)[::-1][:3]
    print(f"\nText {i+1}: {texts[i]}")
    for idx in top_idx:
        print(f"   {feature_names[idx]:<15} {row[idx]:.3f}")
```

```python
#Q4)

import re
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from sklearn.metrics.pairwise import cosine_similarity

# Two short texts on different technologies
text1 = """Cloud computing delivers computing services such as storage, servers, databases and software over the internet.
Companies rent resources on demand instead of buying expensive hardware.
It scales easily, so businesses can handle more data and users when needed.
Providers like AWS and Azure manage security and maintenance for customers.
Cloud computing is widely used for hosting websites, storing data and running applications."""

text2 = """Quantum computing uses qubits that can exist in multiple states at the same time through superposition.
Entanglement links qubits so they can work together in powerful ways.
Quantum computers can solve certain problems, like cryptography and molecular simulation, much faster than classical computers.
The hardware is very delicate and must be kept at extremely low temperatures.
Quantum computing is still experimental and mostly used in research labs."""

# 1. Preprocess and tokenize
def preprocess(text):
    text = text.lower()                          # lowercase
    text = re.sub(r'[^a-z\s]', ' ', text)        # remove punctuation/numbers
    tokens = text.split()                        # tokenize
    return [t for t in tokens if t not in ENGLISH_STOP_WORDS and len(t) > 1]

tok1, tok2 = preprocess(text1), preprocess(text2)
print("Tokens 1:", tok1)
print("Tokens 2:", tok2)

# 2a. Jaccard similarity using sets
set1, set2 = set(tok1), set(tok2)
intersection = set1 & set2
union = set1 | set2
jaccard = len(intersection) / len(union)
print("\nCommon words:", intersection)
print("Jaccard Similarity:", round(jaccard, 4))

# 2b. Cosine similarity using TfidfVectorizer + cosine_similarity()
clean_texts = [" ".join(tok1), " ".join(tok2)]
tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(clean_texts)
cos_sim = cosine_similarity(matrix[0], matrix[1])[0][0]
print("Cosine Similarity:", round(cos_sim, 4))
```

```python
#Q5)


import pandas as pd
from textblob import TextBlob
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Short reviews for a product (smartphone)
reviews = [
    "This smartphone is amazing! The camera is excellent and the battery lasts all day.",
    "Absolutely love this phone. The screen is beautiful and the performance is fantastic.",
    "Great value for money. Fast, smooth and very reliable, I am really happy with it.",
    "Terrible experience. The phone is slow, the battery drains quickly and it keeps crashing.",
    "Worst purchase ever. The screen cracked easily and customer support was awful.",
    "The phone arrived on Monday. It comes with a charger and a box.",
    "The device has a six inch display and 128 GB of storage.",
]

# 1 & 2. Polarity, subjectivity, and classification
rows = []
for r in reviews:
    s = TextBlob(r).sentiment
    if s.polarity > 0.1:
        label = "Positive"
    elif s.polarity < -0.1:
        label = "Negative"
    else:
        label = "Neutral"
    rows.append({"Review": r[:45] + "...", "Polarity": round(s.polarity, 3),
                 "Subjectivity": round(s.subjectivity, 3), "Sentiment": label})

df = pd.DataFrame(rows)
print(df.to_string())
print(df["Sentiment"].value_counts())

# 3. Word cloud for positive reviews only
pos_text = " ".join(r for r, row in zip(reviews, rows) if row["Sentiment"] == "Positive")
wc = WordCloud(width=800, height=400, background_color="white",
               stopwords=STOPWORDS).generate(pos_text)

plt.figure(figsize=(10, 5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud - Positive Reviews")
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# --- Step 1: Choose paragraph and tokenize text ---
paragraph = """
Artificial intelligence and machine learning are transforming the world around us at an incredible pace. 
From healthcare to finance, smart algorithms help automate complex tasks and uncover hidden patterns in large datasets. 
Deep learning models, a specialized subset of neural networks, excel at processing unstructured data like images, audio, and natural language. 
As computational power increases and algorithms become more sophisticated, the boundary between human capability and machine potential continues to blur. 
Building intelligent systems requires careful data preparation, robust training procedures, and thoughtful ethical consideration.
"""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([paragraph])
total_words = len(tokenizer.word_index) + 1

print(f"Total unique words in vocabulary: {total_words}")

# --- Step 2: Create input sequences ---
input_sequences = []
for line in paragraph.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

# Pad sequences so they are all the same length
max_sequence_len = max([len(seq) for seq in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre'))

# Create predictors and label
xs = input_sequences[:, :-1]
labels = input_sequences[:, -1]
ys = tf.keras.utils.to_categorical(labels, num_classes=total_words)

# --- Step 2 (Cont.): Build a simple LSTM model ---
model = Sequential([
    Embedding(total_words, 64, input_length=max_sequence_len-1),
    LSTM(100),
    Dense(total_words, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

# --- Step 3: Train the model ---
history = model.fit(xs, ys, epochs=100, verbose=1)

# --- Step 3 (Cont.): Generate new text starting from a seed word ---
seed_text = "Artificial intelligence"
next_words = 15

for _ in range(next_words):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')
    
    # Predict the probability of the next word
    predicted_probs = model.predict(token_list, verbose=0)
    predicted_index = np.argmax(predicted_probs, axis=-1)[0]
    
    # Find the word corresponding to the index
    output_word = ""
    for word, index in tokenizer.word_index.items():
        if index == predicted_index:
            output_word = word
            break
            
    seed_text += " " + output_word

print("\n--- Generated Text ---")
print(seed_text)
```
